```python
import pandas as pd
import numpy as np

raw_data = {
    'Transaction_ID': ['TXN101', 'TXN102', 'TXN103', 'TXN104', 'TXN104', 'TXN105', 'TXN106', 'TXN107'],
    'Customer_ID': ['CUST_A', 'CUST_B', np.nan, 'CUST_D', 'CUST_D', 'CUST_E', 'CUST_F', 'CUST_G'],
    'Amount': ['150.50', '200.00', '50.25', '120.00', '120.00', np.nan, '310.75', '45.00'],
    'Transaction_Date': ['2026-01-10', '2026-01-11', '2026-01-12', '2026-01-13', '2026-01-13', '2026-01-14', '2026-01-15', '2026-01-16'],
    'Is_Fraud': [0, 0, 1, 0, 0, 0, 1, np.nan]
}

df = pd.DataFrame(raw_data)

print("=== RAW DATASET ===")
print(df)
print("\nData Info Before Cleaning:")
print(df.info())

print("\n--- Missing Values Count ---")
print(df.isnull().sum())

print("\n--- Duplicate Records Count ---")
print(f"Duplicates: {df.duplicated().sum()}")

df = df.drop_duplicates()
df['Customer_ID'] = df['Customer_ID'].fillna('UNKNOWN')
df['Amount'] = df['Amount'].fillna(df['Amount'].astype(float).median())
df['Is_Fraud'] = df['Is_Fraud'].fillna(0)
df['Amount'] = df['Amount'].astype(float)
df['Is_Fraud'] = df['Is_Fraud'].astype(int)
df['Transaction_Date'] = pd.to_datetime(df['Transaction_Date'])

print("\n=== CLEANED DATASET ===")
print(df)
print("\nData Info After Cleaning:")
print(df.info())

output_filename = 'cleaned_dataset.csv'
df.to_csv(output_filename, index=False)
print(f"\nCleaned dataset saved successfully as '{output_filename}'")
```
