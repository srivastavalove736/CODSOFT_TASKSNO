```python
import pandas as pd
import numpy as np

np.random.seed(42)
data = {
    'Transaction_ID': [f'TXN{1000 + i}' for i in range(100)],
    'Customer_Age': np.random.randint(18, 65, size=100),
    'Purchase_Amount': np.round(np.random.normal(loc=150, scale=50, size=100), 2),
    'Product_Category': np.random.choice(['Electronics', 'Clothing', 'Home', 'Books'], size=100),
    'Satisfaction_Score': np.random.randint(1, 6, size=100)
}

data['Purchase_Amount'][5] = 850.00
data['Purchase_Amount'][42] = 920.00

df = pd.DataFrame(data)

print("=== DATASET PREVIEW ===")
print(df.head())
print("\n=== DESCRIPTIVE STATISTICS ===")
print(df.describe())

print("\n=== CATEGORICAL SUMMARY ===")
print(df['Product_Category'].value_counts())

Q1 = df['Purchase_Amount'].quantile(0.25)
Q3 = df['Purchase_Amount'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Purchase_Amount'] < lower_bound) | (df['Purchase_Amount'] > upper_bound)]

print(f"\n=== OUTLIER DETECTION ===")
print(f"Purchase Amount IQR: {IQR:.2f}")
print(f"Detected Outliers ({len(outliers)} total):")
print(outliers[['Transaction_ID', 'Product_Category', 'Purchase_Amount']])

avg_spend_by_category = df.groupby('Product_Category')['Purchase_Amount'].mean().round(2)
total_sales_by_category = df.groupby('Product_Category')['Purchase_Amount'].sum().round(2)
avg_age = df['Customer_Age'].mean()

print("\n=== BUSINESS INSIGHTS ===")
print("Average Spend by Category:")
print(avg_spend_by_category)

print("\nTotal Revenue by Category:")
print(total_sales_by_category)

report = f"""
==================================================
           EXPLORATORY DATA ANALYSIS REPORT
==================================================
1. Total Transactions Analyzed: {len(df)}
2. Average Customer Age: {avg_age:.1f} years
3. Total Revenue Generated: ${df['Purchase_Amount'].sum():,.2f}
4. Highest Revenue Category: {total_sales_by_category.idxmax()} (${total_sales_by_category.max():,.2f})
5. Outliers Detected: {len(outliers)} high-value transaction(s) exceeding ${upper_bound:.2f}
==================================================
"""

print(report)

with open('eda_summary_report.txt', 'w') as f:
    f.write(report)

print("Summary report saved as 'eda_summary_report.txt'")
```
